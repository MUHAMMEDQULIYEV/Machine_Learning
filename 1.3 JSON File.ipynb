The functions `.dump()` and `.dumps()`. 

json module in Python module provides a method called dump() which converts the Python objects into appropriate json objects. It is a slight variant of dumps() method.

>dump() | dumps()
>---|---
>The dump() method is used when the Python objects have to be stored in a file. |The dumps() is used when the objects are required to be in string format and is used for parsing, printing, etc, 
>The dump() needs the json file name in which the output has to be stored as an argument.| The dumps() does not require any such file name to be passed
>This method writes in the memory and then command for writing to disk is executed separately|This method directly writes to the json file
>Faster method | 2 times slower

````python
with open("newjsonfile.json","w") as fobj:
    json.dump(data,fobj)
fobj.close()
````

## Read JSON

Easy to load a JSON object in Python. 
Python has a built-in package called json, which can be used to work with JSON data. 
It’s done by using the json module, which provides us with a lot of methods which among loads() and load() methods are gonna help us to read the JSON file.

## Read JSON file

Let us read the file we just create it. Using the function `load()`. 

````python
import json
with open("newjsonfile.json","r") as fobj:
  jsondata=json.load(fobj)
  print(type(jsondata))
  print(jsondata)
fobj.close()
````

### Deserialization of JSON

The Deserialization of JSON means the conversion of JSON objects into their respective Python objects. 
The load()/loads() method is used for it. 
If you have used JSON data from another program or obtained as a string format of JSON, then it can easily be deserialized with load()/loads(), which is usually used to load from string, otherwise the root object is in list or dict. See the following table given below.

> JSON Object | Python Object 
> --- | ---
> object | dict
> array | list
> string | str
> null | None
> number (int) | int
> number (real) | float 
> true | True
> false | Fase

`json.load()` accepts file object, parses the JSON data, populates a Python dictionary with the data and returns it back to you.

````python
# Python program to read JSON file 
#import json 

fobj = open('data.json') # Opening JSON file 

data = json.load(fobj)# JSON object as a dictionary 

for i in data['students']: # Iterating through the json list 
    print(i) 
  
# Closing file 
fobj.close()
````

## Dictionaries to JSON

The most used process is to save dictionaries to JSON files. 

Dictionary to json file

````python
# python object(dictionary)
dict1 ={ 
    "emp1": { 
        "name": "Samy", 
        "designation": "neuroscientist", 
        "age": "33", 
        "salary": "54000"
    }, 
    "emp2": { 
        "name": "Bachar", 
        "designation": "computerscientist", 
        "age": "28", 
        "salary": "40000"
    }, 
} 

fobj=open("newjsonfile.json","w")
json.dump(dict1, fobj, indent = 6)
fobj.close()
````

Json file to dictionary

````python
with open("newjsonfile.json","r") as fobj:
  dict2=json.load(fobj)
  print(type(dict2))
  print(dict2)
fobj.close()
````

````python
#print(dict2.keys())
#print(dict2.values())

for key in dict2:
  for key2 in dict2[key]:
    print(key, key2, dict2[key][key2])
````
